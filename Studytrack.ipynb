```python
tasks = []


def show_tasks():
  #سوينا داله ووظيفتها عرض المهام
    print("\n===== MY STUDY PLAN =====")

    if not tasks:
        print("No study tasks yet.")
        return

    for i, task in enumerate(tasks, start=1):
      #هنا نمر على المهام واحده بوحده
        status = "✓" if task["done"] else "○"
        print(f"{i}. {status} {task['subject']} - {task['task']} ({task['hours']}h)")


def add_task():
    print("\n--- Add New Task ---")

    subject = input("Subject: ")
    task_name = input("Task: ")

    while True:
        try:
            hours = float(input("Study hours: "))

            if hours <= 0:
                print("Please enter a number greater than 0.")
            else:
                break

        except ValueError:
            print("Please enter a valid number.")

    tasks.append({
        "subject": subject,
        "task": task_name,
        "hours": hours,
        "done": False
    })

    print("Task added successfully!")


def mark_task_done():
    show_tasks()

    if not tasks:
        return

    try:
        number = int(input("\nEnter task number to mark as done: "))

        if 1 <= number <= len(tasks):
            tasks[number - 1]["done"] = True
            print("Task completed!")
        else:
            print("Invalid task number.")

    except ValueError:
        print("Please enter a valid number.")


def show_statistics():
  #داله تعرض احصائيات الدراسه
    if not tasks:
        print("\nNo tasks available.")
        return

    total_hours = sum(task["hours"] for task in tasks)
    completed = sum(task["done"] for task in tasks)

    percentage = (completed / len(tasks)) * 100

    print("\n===== STUDY STATISTICS =====")
    print("Total tasks:", len(tasks))
    print("Completed:", completed)
    print("Remaining:", len(tasks) - completed)
    print("Total study hours:", total_hours)
    print("Progress:", round(percentage, 1), "%")


while True:
    print("\n============================")
    print("       📚 STUDYTRACK")
    print("============================")
    print("1. Show study plan")
    print("2. Add task")
    print("3. Mark task as done")
    print("4. Show statistics")
    print("5. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        show_tasks()

    elif choice == "2":
        add_task()

    elif choice == "3":
        mark_task_done()

    elif choice == "4":
        show_statistics()

    elif choice == "5":
        print("Good luck with your studies!")
        break

    else:
        print("Invalid choice. Please try again.")
```
